```python
#@title 1 · Prepare BreezeSprint25 v0.1.1
#@markdown **First:** Runtime → Change runtime type → TPU v5e-1 (free quota permitting).
#@markdown 臺灣華語與中英混用。第二格上傳，完成後自動下載時間戳 TXT；不需 API key。
# Full sources: https://github.com/thc1006/breezesprint-25 · Independent integration, not an official Google/MediaTek/NYCU release.
# This standalone notebook embeds the same readable source files as the repository. No source downloads.
import base64, hashlib, importlib, io, json, sys, zipfile
from pathlib import Path

if "sprint" in globals():
    sprint.close()
# Generated by tools/build_notebook.py; the payload contains Python SOURCE, not model weights.
# Its digest detects corruption, not the trustworthiness of the publisher.
PAYLOAD_SHA256 = "449e8908e6cda6c7a828a1fa064db8c2a88b3a622fd6458e0f2a40fa05ea93b6"
PAYLOAD_FILES = ['src/breezesprint25/__init__.py', 'src/breezesprint25/assets.py', 'src/breezesprint25/checkpoint.py', 'src/breezesprint25/colab.py', 'src/breezesprint25/core.py', 'src/breezesprint25/engine.py', 'src/breezesprint25/frontend.py', 'src/breezesprint25/loading.py', 'src/breezesprint25/pipeline.py', 'src/breezesprint25/preflight.py', 'src/breezesprint25/text_output.py', 'src/breezesprint25/vocabulary.py', 'src/breezesprint25/worker.py', 'src/breezesprint25/model.json', 'requirements-tpu.lock']
PAYLOAD_B64 = ""
raw = base64.b64decode(PAYLOAD_B64, validate=True)
if hashlib.sha256(raw).hexdigest() != PAYLOAD_SHA256:
    raise RuntimeError("Notebook source payload checksum mismatch. Download a clean copy.")
bundle = Path("/content/breezesprint25/app")
bundle.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(raw)) as z:
    if sorted(z.namelist()) != sorted(PAYLOAD_FILES):
        raise RuntimeError("Unexpected notebook payload files.")
    for info in z.infolist():
        dest = (bundle/info.filename).resolve()
        if not dest.is_relative_to(bundle.resolve()) or info.file_size > 2_000_000:
            raise RuntimeError("Unsafe notebook payload path or size.")
        dest.parent.mkdir(parents=True, exist_ok=True)
        dest.write_bytes(z.read(info.filename))
# Colab's kernel imports only the stdlib controller. JAX stays in the isolated worker.
for key in list(sys.modules):
    if key == "breezesprint25" or key.startswith("breezesprint25."):
        del sys.modules[key]
sys.path.insert(0, str(bundle/"src"))
from breezesprint25.colab import setup
sprint = setup(bundle)
del raw
```
